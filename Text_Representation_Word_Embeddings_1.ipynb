```python
import pandas as pd
from nltk.tokenize import sent_tokenize
from gensim.utils import simple_preprocess

# Read CSV
df = pd.read_csv("battles.csv")

# Check columns
print(df.columns)

# Create empty list
story = []

# Suppose text column is named 'name'
# Replace 'name' with your actual text column

for text in df['name'].dropna():

    # Sentence tokenization
    raw_sent = sent_tokenize(str(text))

    # Word tokenization + preprocessing
    for sent in raw_sent:
        story.append(simple_preprocess(sent))

# View output
print(story[:5])
```

```python
story[0]
```

```python
model = gensim.models.Word2Vec(
    window=10,#Look 10 words left and 10 words right
    min_count=2#Ignore rare words appearing less than 2 times.
)
```

```python
#Word2Vec scans all words,and builds:unique word list,frequency count,indexing
x=model.build_vocab(story)
x
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.wv.most_similar('battle')
```

```python
model.wv.similarity('battle','cailin')
```
