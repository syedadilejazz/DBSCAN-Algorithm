Bag of words(Sklearn CountVectorizer)

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({"text":["people watch dswithbappy",
                         "dswithbappy watch dswithbappy",
                         "people write comment",
                          "dswithbappy write comment"],"output":[1,1,0,0]})

df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
bow = cv.fit_transform(df['text'])
bow
```

```python
#vocabulary
print(cv.vocabulary_)
```

```python
bow.toarray()
```

```python
print(df["text"][0])
print(bow[0].toarray())
```

```python
print(df["text"][1])
print(bow[1].toarray())
```

```python
print(df["text"][2])
print(bow[2].toarray())
```

```python
print(df["text"][3])
print(bow[3].toarray())
```

```python
df = pd.DataFrame({"text":["people watch dswithbappy",
                         "dswithbappy watch dswithbappy",
                         "people write comment",
                          "dswithbappy write comment"],"output":[1,1,0,0]})

df
```

```python
bow = cv.fit_transform(df['text'])
```

```python
print(cv.vocabulary_)
```

```python
print(df["text"][0])
print(bow[0].toarray())
```

Important word

```python
df = pd.DataFrame({"text":["people watch dswithbappy",
                         "dswithbappy watch dswithbappy",
                         "people write comment",
                          "dswithbappy write comment"],"output":[1,1,0,0]})

df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfid= TfidfVectorizer()
```

```python
arr = tfid.fit_transform(df['text']).toarray()
arr
```

```python
print(tfid.idf_)
```

Cosine Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity

# Compute cosine similarity
similarity = cosine_similarity(arr)

print(similarity)
```

```python
import pandas as pd

similarity_matrix = cosine_similarity(arr)
docs = ['D1', 'D2', 'D3', 'D4']

df_sim = pd.DataFrame(
    similarity_matrix,
    columns=docs,
    index=docs
)
print(df_sim)
```
